```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")

api_key = os.getenv("GOOGLE_API_KEY")
if not api_key:
    raise ValueError("GOOGLE_API_KEY not found in .env file")

print("API key loaded successfully!")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    google_api_key=api_key,
    temperature=0,
    max_retries=1,
    timeout=20
)
```

```python
try:
    response = model.invoke("Explain AI agents in two sentences.")
    print(response.content)

except Exception as e:
    print("Error type:", type(e).__name__)
    print("Error details:", str(e))
```

```python
from langchain.agents  import create_agent
from langchain.tools import tool

@tool
def get_weather(location:str)->str:
    """Get the waether at a location"""
    return f"It's sunnay in {location}"

agent=create_agent(
    model="gemini-1.5-flash",
    tools=[get_weather],
    system_prompt="You are helpful assistant."
)
agent
```

```python
### We provide the message in the form of the dictionary
try:
    response = agent.invoke({"messages": [{"role": "user", "content": "What is the weather like in New York?"}]})
    print(response)
except Exception as e:
    print("Error type:", type(e).__name__)
    print("Error details:", str(e))
```
