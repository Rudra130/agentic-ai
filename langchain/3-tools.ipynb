```python
import os
from langchain_nvidia_ai_endpoints import ChatNVIDIA
from IPython.display import display, Markdown

model = ChatNVIDIA(
  base_url="https://integrate.api.nvidia.com/v1",
  model="nvidia/nemotron-3.5-super-vl-120b-a12b",
  api_key=os.environ["NVIDIA_API_KEY"],
  temperature=1,
  top_p=0.95,
  max_completion_tokens=16384,
  seed=42,
)

try:
  response= model.invoke("My name is rudra , So tell me what is my name")
  print(display(response.content))
except Exception as e:
    print("Error type:", type(e).__name__)
    print("Error details:", str(e))
```

## Tool calling 

```python
from langchain.agents import create_agent
from langchain.tools import tool

@tool
def get_weather(city: str) -> str:
    """Get the weather report of the city"""
    return f"The weather of the {city} is sunny"


agent = create_agent(
    model=model,
    tools=[get_weather],
    system_prompt="You are helpful assistant."
)

try:
    response = agent.invoke({"messages": [{"role": "user", "content": "What is the weather like in New York?"}]})
    print(response["messages"][-1].content)
except Exception as e:
    print("Error type:", type(e).__name__)
    print("Error details:", str(e))
```
